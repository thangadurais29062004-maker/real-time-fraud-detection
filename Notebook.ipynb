```python
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

# 1. Load data and define X and y
df = pd.read_csv("fraud_dataset.csv")   # use your actual file name/path
X = df.drop(columns=["is_fraud", "transaction_id"])
y = df["is_fraud"]

# 2. Column groups
cat_cols = ["merchant_category", "location", "device_type"]
num_cols = ["amount", "transaction_hour", "transaction_frequency", "previous_fraud_count"]

def make_preprocessor(scale=False):
    return ColumnTransformer([
        ("cat", OneHotEncoder(handle_unknown="ignore"), cat_cols),
        ("num", StandardScaler() if scale else "passthrough", num_cols),
    ])

neg, pos = (y == 0).sum(), (y == 1).sum()

# 3. The three pipelines
lr_pipeline = Pipeline([
    ("prep", make_preprocessor(scale=True)),
    ("model", LogisticRegression(class_weight="balanced", max_iter=1000)),
])

rf_pipeline = Pipeline([
    ("prep", make_preprocessor()),
    ("model", RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42)),
])

xgb_pipeline = Pipeline([
    ("prep", make_preprocessor()),
    ("model", XGBClassifier(scale_pos_weight=neg / pos, eval_metric="logloss", random_state=42)),
])
```

```python
import pandas as pd
from sklearn.model_selection import StratifiedKFold, cross_val_predict
from sklearn.metrics import roc_auc_score, average_precision_score, precision_score, recall_score, f1_score

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
models = {"LogReg": lr_pipeline, "RandomForest": rf_pipeline, "XGBoost": xgb_pipeline}

rows = []
for name, m in models.items():
    proba = cross_val_predict(m, X, y, cv=cv, method="predict_proba")[:, 1]
    pred = (proba >= 0.5).astype(int)
    rows.append({
        "model": name,
        "ROC-AUC": roc_auc_score(y, proba),
        "PR-AUC": average_precision_score(y, proba),
        "precision": precision_score(y, pred),
        "recall": recall_score(y, pred),
        "F1": f1_score(y, pred),
    })

print(pd.DataFrame(rows).round(3))
```

```python
import numpy as np
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import precision_score, recall_score, f1_score

proba_xgb = cross_val_predict(xgb_pipeline, X, y, cv=cv, method="predict_proba")[:, 1]

rows = []
for t in np.arange(0.10, 0.95, 0.05):
    pred = (proba_xgb >= t).astype(int)
    rows.append({
        "threshold": round(t, 2),
        "precision": precision_score(y, pred, zero_division=0),
        "recall": recall_score(y, pred),
        "F1": f1_score(y, pred),
    })

print(pd.DataFrame(rows).round(3))
```

```python
scores = pd.Series(proba_xgb, index=y.index)
zone = pd.cut(scores, bins=[0, 0.30, 0.80, 1.0],
              labels=["ALLOW", "ALERT", "BLOCK"], right=False, include_lowest=True)

summary = pd.DataFrame({"zone": zone, "is_fraud": y}).groupby("zone").agg(
    transactions=("is_fraud", "size"),
    frauds=("is_fraud", "sum"),
    fraud_rate=("is_fraud", "mean"),
)
print(summary.round(3))
```
